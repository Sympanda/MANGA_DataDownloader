```python
import json
import os
import numpy as np
import matplotlib.pyplot as plt

def load_sample(index, folder="mmu_huggingface_individual"):
    path = os.path.join(folder, f"sample_{index:03d}.json")
    with open(path, "r") as f:
        return json.load(f)

sample = load_sample(0)
sample.keys()
```

```python
# Need to check the flux values for extreme outliers
# Flux values should be between 0 and 1. Treat extrmes as 0
# Plot images with this in mind

for i, image in enumerate(sample["images"]):
    flux_data = np.array(image['flux'])
    # set extreme values to 0
    flux_data_clipped = np.clip(flux_data, 0, 1)  # Clip values to range [0, 1]
    
    plt.figure(figsize=(8, 6))
    plt.imshow(flux_data_clipped, origin="lower", cmap="gray")
    plt.title(f"Image {i} ({image['filter']}) - Clipped Flux")
    plt.colorbar(label="Flux (clipped)")
    plt.axis("off")
    plt.show()
```

```python
# Can we dynamically clip the flux values based on the histogram?
# If a flux value is outside the 95th percentile, set it to 0

def adaptive_clip(image, lower_q=0.25, upper_q=99.75):
    lower = np.percentile(image, lower_q)
    upper = np.percentile(image, upper_q)
    return np.clip(image, lower, upper)


# Apply clipping to each image's flux data
for i, image in enumerate(sample["images"]):
    flux_data = np.array(image['flux'])
    clipped_flux_data = adaptive_clip(flux_data)
    
    plt.figure(figsize=(8, 6))
    plt.imshow(clipped_flux_data, origin="lower", cmap="gray")
    plt.title(f"Image {i} ({image['filter']}) - Clipped Flux (95th Percentile)")
    plt.colorbar(label="Flux (clipped)")
    plt.axis("off")
    plt.show()
```

```python
import numpy as np

def clean_edge_ceiling_pixels(image, r_frac=0.38):
    h, w = image.shape
    cy, cx = h // 2, w // 2
    yy, xx = np.ogrid[:h, :w]
    rr = np.sqrt((yy - cy)**2 + (xx - cx)**2)
    r_max = np.max(rr)
    print(f"Image shape: {image.shape}, Center: ({cy}, {cx}), Max radius: {r_max}")
    
    # 1. Compute clipping bounds
    lower_clip = np.percentile(image, 0.25)
    upper_clip = np.percentile(image, 99.75)
    clipped = np.clip(image, lower_clip, upper_clip)
    
    # 2. Create mask for far-from-center (e.g., outer 20%)
    far_mask = rr > (r_frac * r_max)
    print(f"Far mask shape: {far_mask.shape}, Far pixels count: {np.sum(far_mask)}")
    
    # 3. Create mask for "ceiling" values
    upper = np.percentile(image, 97.5)
    near_ceiling = (image - upper) >= 0  # adjust tolerance if needed
    
    # 4. Combine masks
    junk_mask = far_mask & near_ceiling
    
    # 5. Replace junk values with background (lower clip value)
    cleaned = clipped.copy()
    cleaned[junk_mask] = lower_clip
    
    return cleaned, junk_mask


# Apply the cleaning function to each image's flux data
for i, image in enumerate(sample["images"]):
    flux_data = np.array(image['flux'])
    cleaned_flux_data, junk_mask = clean_edge_ceiling_pixels(flux_data)
    
    plt.figure(figsize=(12, 6))
    
    # Original image
    plt.subplot(1, 2, 1)
    plt.imshow(flux_data, origin="lower", cmap="gray")
    plt.title(f"Original Image {i} ({image['filter']})")
    plt.colorbar(label="Flux")
    plt.axis("off")
    
    # Cleaned image
    plt.subplot(1, 2, 2)
    plt.imshow(cleaned_flux_data, origin="lower", cmap="gray")
    plt.title(f"Cleaned Image {i} ({image['filter']})")
    plt.colorbar(label="Flux (cleaned)")
    plt.axis("off")
    
    plt.tight_layout()
    plt.show()
```
